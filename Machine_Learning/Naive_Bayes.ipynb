```python
from sklearn.datasets import load_iris, load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, precision_score,recall_score
from sklearn.preprocessing import StandardScaler
```

```python
#1.IRIS Dataset
# GaussianNB
X=load_iris().data
y=load_iris().target
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)
model=GaussianNB()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
scale=StandardScaler()
X_train_scale=scale.fit_transform(X_train)
X_test_scale=scale.transform(X_test)
```

```python
model=GaussianNB()
model.fit(X_train_scale,y_train)
y_pred=model.predict(X_test_scale)
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
#2.Breast Cancer 
# GaussianNB
data=load_breast_cancer()
X=data.data
y=data.target
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)
model=GaussianNB()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
print(accuracy_score(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

```python
print(precision_score(y_pred,y_test))
print(recall_score(y_pred,y_test))
```

```python
print(classification_report(y_pred,y_test))
```

```python
#3.Email Spam Detection
# MultinomialNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer

emails=[
    "free cash available",
    "let me know",
    "apple phone available for free",
    "we will meet tommorrow",
    "cash credit to you"
]

labels=[1,0,1,0,1]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)


model=MultinomialNB()
model.fit(X,labels)
text=["we will meet today"]
text_mail=vectorizer.transform(text)
y_pred=model.predict(text_mail)
print(y_pred)

text=["free cash available"]
text_mail=vectorizer.transform(text)
y_pred=model.predict(text_mail)
print(y_pred)
```

```python
# 4.Sentiment Analysis
# MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
reviews = [
    "movie is awesome",
    "excellent acting",
    "bad movie",
    "worst film ever"
]

sentiment = [1, 1, 0, 0]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(reviews)
y = sentiment
model = MultinomialNB()
model.fit(X,y)
text = ["excellent movie"]
review_text = vectorizer.transform(text)
y_pred = model.predict(review_text)
print(y_pred)
```

```python
# 5.Naive Bayes
# BernoulliNB
from sklearn.naive_bayes import BernoulliNB
X = [
    [1, 1], #has Job, has house
    [1, 0],
    [0, 1],
    [0, 0],
    [1, 1],
    [0, 0]
]

y = [1, 1, 0, 0, 1, 0]
model = BernoulliNB()
model.fit(X, y)
prediction = model.predict([[1, 1]])

print("Prediction:", prediction)
```
